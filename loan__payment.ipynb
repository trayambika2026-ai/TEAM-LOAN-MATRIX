```python
import numpy as np
import matplotlib.pyplot as plt
loan = float(input("Enter loan amount: "))
annual_rate = float(input("Enter annual interest rate (%): "))
payment = float(input("Enter monthly payment: "))
months = int(input("Enter number of months: "))

r = annual_rate / 100 / 12

print("\nMonthly interest rate =", r)
```

```python
r = annual_rate / 100 / 12
```

```python
A = np.array([
    [1 + r, -payment],
    [0, 1]
])

x0 = np.array([
    [loan],
    [1]
])

print("Transition Matrix A:")
print(A)

print("\nInitial State Vector x0:")
print(x0)
```

```python
balances = [loan]

for n in range(1, months + 1):
    xn = np.linalg.matrix_power(A, n) @ x0
    balance = xn[0, 0]

    # A loan cannot have a negative remaining balance
    balance = max(0, balance)

    balances.append(balance)

print("Month-by-month loan balance:")
for month, balance in enumerate(balances):
    print(f"Month {month}: {balance:.2f}")
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    range(len(balances)),
    balances,
    marker="o"
)

plt.xlabel("Month")
plt.ylabel("Remaining Loan Balance")
plt.title("Loan Balance Over Time")

plt.grid(True)
plt.show()
```

```python
eigenvalues = np.linalg.eigvals(A)

print("Eigenvalues:")
for value in eigenvalues:
    print(f"{value:.4f}")
```

```python
trace_A = np.trace(A)
det_A = np.linalg.det(A)

print("Trace(A) =", trace_A)
print("Determinant(A) =", det_A)

print("\nCayley-Hamilton equation:")
print(f"A^2 - ({trace_A:.4f})A + ({det_A:.4f})I = 0")
```

```python
I = np.eye(2)

left_side = (
    np.linalg.matrix_power(A, 2)
    - trace_A * A
    + det_A * I
)

print("Cayley-Hamilton verification:")
print(left_side)
```
